```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv
```

```python
load_dotenv()

API_KEY = os.getenv("HF_TOKEN")
MODEL_NAME = os.getenv("HF_MODEL")

print("Using model:", MODEL_NAME)
```

```python
url = "https://router.huggingface.co/v1/models"

headers = {
    "Authorization": f"Bearer {API_KEY}"
}

response = requests.get(url, headers=headers)
models = response.json()

print("Total models available:", len(models["data"]))
```

```python
model_data = []

for m in models["data"]:
    model_data.append({
        "Model Name": m["id"],
        "Provider": m.get("owned_by", "Unknown")
    })

df_models = pd.DataFrame(model_data)
df_models.head(20)
```

```python
df_models[df_models["Model Name"] == MODEL_NAME]
```

```python
API_URL = "https://router.huggingface.co/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}
```

```python
default_prompt = "Explain artificial intelligence in simple words."

payload = {
    "model": MODEL_NAME,
    "messages": [
        {"role": "user", "content": default_prompt}
    ],
    "max_tokens": 1024,
    "temperature": 0.7,
    "top_p": 0.9
}

response = requests.post(API_URL, headers=headers, json=payload)

result = response.json()

print("Assistant:\n")
print(result["choices"][0]["message"]["content"])
```

```python
usage = result.get("usage", {})

prompt_tokens = usage.get("prompt_tokens", 0)
completion_tokens = usage.get("completion_tokens", 0)
total_tokens = usage.get("total_tokens", 0)

CONTEXT_LIMIT = 32768
remaining_tokens = CONTEXT_LIMIT - total_tokens
usage_percent = (total_tokens / CONTEXT_LIMIT) * 100

print("\nToken Usage Summary")
print("---------------------")

print(f"Prompt Tokens       : {prompt_tokens}")
print(f"Completion Tokens   : {completion_tokens}")
print(f"Total Tokens Used   : {total_tokens}")
print(f"Remaining Tokens    : {remaining_tokens}")
print(f"Context Used        : {usage_percent:.2f}%")
```
